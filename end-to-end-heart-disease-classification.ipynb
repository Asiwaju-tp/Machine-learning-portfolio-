## Features
This is where you'll get different information about each of the features in yout data

**Create data dictionary**

1. age (Age of the patient in years)
2. sex (Male = 1/Female = 0)
3. cp chest pain type ([typical angina, atypical angina, non-anginal, asymptomatic])
   * 0: Typical angina: chest pain related to decrease blood supply to the heart
   * 1: Atypical angina: chest pain not related to the heart
   * 2: Non-angina pain: typically esophageal spasms(non heart related)
   * 3: Asymptomatic: chest pain not showing a sign of disease

4. trestbps resting blood pressure (resting blood pressure (in mm Hg on admission to the hospital)) anything above 130-140 is typically cause for concern
   
5. chol (serum cholesterol in mg/dl)
   * serum = LDL + HDL + .2* triglycerides
   * above 200 is cause for concern
     
6. fbs (if fasting blood sugar > 120 mg/dl) (1=true; 0=False)
   * '>126' mg/dL signals diabetes

7. restecg (resting electrocardiographic results)
   * 0: Nothing to note
   * 1: ST-T Wave abnormality

8. thalach: maximum heart rate achieved
9. exang: exercise-induced angina (Yes=1; No=0)
10. oldpeak: ST depression induced by exercise relative to rest looks at stress of heart during exercise unhealthy heart will be stress more
11. slope: the slope of the peak exercise ST segment
    * 0:Upsloping: better heart rate
    * 1:Flatsloping: minimal change(typical healthy heart)
    * 2:Downsloping: signs of unhealthy heart
      
12. ca: number of major vessels (0-3) colored by fluoroscopy
    * coloured vessels means the doctor can see the blood passing through
    * the more blood movement the better(no clots)
    
13. thal: [normal; fixed defect; reversible defect]
    * 1,3:normal
    * 6: fixed defect: use to be defect but okay now
    * 7: reversable defect: no proper blood movement when exercising

14. target - have disease or not ( 1 or 0)
15. origin (place of study)
16. num: the predicted attribute

Now we've got our data split into training and test set, its time to build a machine learning model

we'll train it and find the pattern on the training set.

And we'll test it (use the patterns) on the test set. 

we're going to try out three(3) differnt machine learning models;
1. Logistic Regression
2. k-Nearest Neighbour Classifier
3. Random Forest Classifier

```python
# Put models in dictionary
models = {"Logistic Regression": LogisticRegression(),
          "KNN": KNeighborsClassifier(),
          "Random Forest": RandomForestClassifier()}

# Create a function to fit and score models
def fit_and_score(models, x_train, x_test, y_train, y_test):
    """
    Fits and evaluate given machine learning models.
    models : a dict of different Scikit-Learning machine learning models.
    x_train : training data (no labels)
    x_test : testing data (no labels)
    y_train : training labels 
    y_test : testing labels
    """
    # Set random seed
    np.random.seed(42)
    # Make a dictionary to keep the model_scores
    model_scores = {}
    # Loop through the models 
    for name, model in models.items():
        # Fit the model to the data
        model.fit(x_train, y_train)
        # Evaluate the model and append its score to the model_scores
        model_scores[name] = model.score(x_test, y_test)
    return model_scores
```

```python
model_scores = fit_and_score(models=models,
                             x_train=x_train,
                             x_test=x_test,
                             y_train=y_train,
                             y_test=y_test)

model_scores
```

## Hperparameter tuning

```python
# we can search google on how to go about it(how to tune a kneighbors classifier model)
#Let's tune KNN
train_scores = []
test_scores = []

# Create a list of different values for n_neighbors
neighbors = range(1, 21)

# Setup KNN instance
knn = KNeighborsClassifier()

# Loop through different n_neighbors
for i in neighbors:
    knn.set_params(n_neighbors=i)

    # Fit the algorithm
    knn.fit(x_train, y_train)

    # Update the training_scores list
    train_scores.append(knn.score(x_train, y_train))

    # Update the test_scores list
    test_scores.append(knn.score(x_test, y_test))
```

```python
train_scores
```

```python
test_scores
```

```python
plt.plot(neighbors, train_scores, label="Train score")
plt.plot(neighbors, test_scores, label="Test score")
# This helps to scale down the graph and number from 1 to 21 with a difference of 1
plt.xticks(np.arange(1, 21, 1))  
plt.xlabel("Number of neighbors")
plt.ylabel("Model score")
plt.legend()


print(f"Maximum KNN score of the test data: {max(test_scores)*100:.2f}%")

plt.show()
```

## Hyperparameter tuning with RandomizedSearchCV

We are going to tune:
* Logistic Regression
* RandomForestClassifier
  
...Using RandomizedSearchCV

```python
# Hyperparameters  for both LogisticRegression and RandomForestClassifier

# Create hyperparameter grid for LogisticRegression
log_reg_grid = {"C": np.logspace(-4, 4, 20),
                "solver": ["liblinear"]}


# Create a hyperparameter grid for RandomForestClassifier
rf_grid = {"n_estimators": np.arange(10, 1000, 50),
           "max_depth": [None, 3, 5, 10],
           "min_samples_split": np.arange(2, 20, 2),
           "min_samples_leaf": np.arange(1, 20, 2)}
```

Now we've got hyperparameter grids setup for each of our model, lets tune them using RandomizedSearchCV...

```python
# Tune LogisticRegression

np.random.seed(42)

# Setup random hyperparameter search for LogisticRegression
rs_log_reg = RandomizedSearchCV(LogisticRegression(),
                                param_distributions=log_reg_grid,
                                cv=5,
                                n_iter=20,
                                verbose=True)

# Fit random hyperparameter search for LogisticRegression
rs_log_reg.fit(x_train, y_train);
```

```python
rs_log_reg.best_params_
```

```python
rs_log_reg.score(x_test, y_test)
```

Now we've tuned LogisticRegression(), let's do the same for RandomClassifier()

```python
# Setup Random seed
np.random.seed(42)

# Setup random hyperparameter search for RandomForestClassifier
rs_rf = RandomizedSearchCV(RandomForestClassifier(),
                           param_distributions = rf_grid,
                           cv=5,
                           n_iter=20,
                           verbose=True)

# Fit random hyperparameter search model for RandomForestClassifier()
rs_rf.fit(x_train, y_train)
```

```python
# Find the best hyperparameters
rs_rf.best_params_
```

```python
# Evaluate the randomized search RandomForestClassifier model
rs_rf.score(x_test, y_test)
```

## Hyperparameter Tuning With GridSearchCV

## Calculate evaluation metrics using cross-validation

we're going to calculate accuracy, precision,recall and F-1 score of our model using cross-validation and to do so we'll be using `cross_val_score().`

```python
# Check the best Hyperparameter

gs_log_reg.best_params_
```

```python
# Create a new classifier with best parameters
clf = LogisticRegression(C=0.20433597178569418,
                         solver="liblinear")
```

```python
# Cross-validated accuracy
cv_acc = cross_val_score(clf,
                         x,
                         y,
                         cv=5,
                         scoring="accuracy")
cv_acc
```

```python
cv_acc = np.mean(cv_acc)
cv_acc
```

```python
# Cross-validated precision
cv_precision = cross_val_score(clf,
                                 x,
                                 y,
                                 cv=5,
                                 scoring="precision")

cv_precision = np.mean(cv_precision)
cv_precision
```

```python
# Cross-validated recall
cv_recall = cross_val_score(clf,
                             x,
                             y,
                             cv=5,
                             scoring="recall")

cv_recall = np.mean(cv_recall)
cv_recall
```

```python
cv_f1 = cross_val_score(clf,
                         x,
                         y,
                         cv=5,
                         scoring="f1")

cv_f1 = np.mean(cv_f1)
cv_f1
```

```python
# Visualizing the cross-validated matrix
cv_metrics = pd.DataFrame({"Accuracy": cv_acc,
                           "Precision": cv_precision,
                           "Recall": cv_recall,
                           "F1": cv_f1},
                          index=[0])

cv_metrics.T.plot.bar(title="Cross-Validated classification metrics",
                      legend=False)

plt.show()
```
